```python
import json

from ROMP_twin import (
    ROMPTwin,
    InitiationChannel,
    PropagationChannel,
    TerminationChannel
)

from ROMP_helpers import initialize_simulation_volume, round_dict

TOTAL_MONOMER_COUNT = 15000
INITIATOR_COUNT = int(TOTAL_MONOMER_COUNT // 50)
TOTAL_MOLARITY = 0.100
MONOMER_NAMES = ["M1"]

monomer_molarities = {"M1": TOTAL_MOLARITY}
solvent_volume, monomer_counts = initialize_simulation_volume(
    total_target_molecules=TOTAL_MONOMER_COUNT, 
    target_molarities=monomer_molarities
)

channels = [
    InitiationChannel(monomer_name="M1", rate_constant=1e03),
    PropagationChannel(chain_end_name="M1", monomer_name="M1", rate_constant=1.0),
    TerminationChannel(chain_end_name="M1", rate_constant=1e-05),
]

twin = ROMPTwin(
    monomer_names=MONOMER_NAMES, monomer_mws={"M1":100.},
    monomer_molar_volumes={"M1":100.},
    polymer_molar_volumes={"M1":98.},
    monomer_counts=monomer_counts, initiator_count=INITIATOR_COUNT,
    solvent_volume=solvent_volume, channels=channels
)

twin.run_to_time(end_time = 60.0)

report = twin.get_report()

print(json.dumps(round_dict(report, decimals=3), indent=4))
```
